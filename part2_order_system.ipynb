```python
# Given menu (keep as it is)
menu = {
    "Paneer Tikka":   {"category": "Starters",  "price": 180.0, "available": True},
    "Chicken Wings":  {"category": "Starters",  "price": 220.0, "available": False},
    "Veg Soup":       {"category": "Starters",  "price": 120.0, "available": True},
    "Butter Chicken": {"category": "Mains",     "price": 320.0, "available": True},
    "Dal Tadka":      {"category": "Mains",     "price": 180.0, "available": True},
    "Veg Biryani":    {"category": "Mains",     "price": 250.0, "available": True},
    "Garlic Naan":    {"category": "Mains",     "price":  40.0, "available": True},
    "Gulab Jamun":    {"category": "Desserts",  "price":  90.0, "available": True},
    "Rasgulla":       {"category": "Desserts",  "price":  80.0, "available": True},
    "Ice Cream":      {"category": "Desserts",  "price": 110.0, "available": False},
}

# ----- Given menu -----
menu = {
    "Paneer Tikka":   {"category": "Starters",  "price": 180.0, "available": True},
    "Chicken Wings":  {"category": "Starters",  "price": 220.0, "available": False},
    "Veg Soup":       {"category": "Starters",  "price": 120.0, "available": True},
    "Butter Chicken": {"category": "Mains",     "price": 320.0, "available": True},
    "Dal Tadka":      {"category": "Mains",     "price": 180.0, "available": True},
    "Veg Biryani":    {"category": "Mains",     "price": 250.0, "available": True},
    "Garlic Naan":    {"category": "Mains",     "price":  40.0, "available": True},
    "Gulab Jamun":    {"category": "Desserts",  "price":  90.0, "available": True},
    "Rasgulla":       {"category": "Desserts",  "price":  80.0, "available": True},
    "Ice Cream":      {"category": "Desserts",  "price": 110.0, "available": False},
}

# Step 1: start with an empty cart
cart = []

# Step 2: add item
def add_to_cart(item_name, quantity):
    if item_name not in menu:
        print(f"Cannot add '{item_name}': item not found in menu.")
        return

    item_info = menu[item_name]

    if not item_info["available"]:
        print(f"Cannot add '{item_name}': item is currently unavailable.")
        return

    # check if already present
    for entry in cart:
        if entry["item"] == item_name:
            entry["quantity"] += quantity
            print(f"Updated '{item_name}' quantity to {entry['quantity']}.")
            return

    # otherwise add new entry
    cart.append({
        "item": item_name,
        "quantity": quantity,
        "price": item_info["price"]
    })
    print(f"Added '{item_name}' x{quantity} to cart.")

# Step 3: remove item
def remove_from_cart(item_name):
    for entry in cart:
        if entry["item"] == item_name:
            cart.remove(entry)
            print(f"Removed '{item_name}' from cart.")
            return
    print(f"'{item_name}' not found in cart.")

# Step 4: update quantity
def update_quantity(item_name, new_qty):
    for entry in cart:
        if entry["item"] == item_name:
            entry["quantity"] = new_qty
            print(f"Quantity for '{item_name}' updated to {new_qty}.")
            return
    print(f"Cannot update: '{item_name}' not in cart.")

# helper: print cart
def show_cart():
    print("\nCurrent Cart:")
    if not cart:
        print("(Cart is empty)")
        return
    for entry in cart:
        line_total = entry["quantity"] * entry["price"]
        print(f"{entry['item']:<18} x{entry['quantity']}   ₹{line_total:.2f}")

# ---------- Simulate required sequence ----------
add_to_cart("Paneer Tikka", 2)
show_cart()

add_to_cart("Gulab Jamun", 1)
show_cart()

add_to_cart("Paneer Tikka", 1)   # quantity should become 3
show_cart()

add_to_cart("Mystery Burger", 1)  # not in menu
add_to_cart("Chicken Wings", 1)   # unavailable

remove_from_cart("Gulab Jamun")
show_cart()

# ---------- Final Order Summary ----------
print("\n========== Order Summary ==========")
subtotal = 0.0

for entry in cart:
    item_total = entry["quantity"] * entry["price"]
    subtotal += item_total
    print(f"{entry['item']:<18} x{entry['quantity']}   ₹{item_total:.2f}")

print("------------------------------------")
gst = subtotal * 0.05
grand_total = subtotal + gst

print(f"Subtotal:                ₹{subtotal:.2f}")
print(f"GST (5%):                ₹{gst:.2f}")
print(f"Total Payable:           ₹{grand_total:.2f}")
print("====================================")
```

```python
menu = {
    "Paneer Tikka":   {"category": "Starters",  "price": 180.0, "available": True},
    "Chicken Wings":  {"category": "Starters",  "price": 220.0, "available": False},
    "Veg Soup":       {"category": "Starters",  "price": 120.0, "available": True},
    "Butter Chicken": {"category": "Mains",     "price": 320.0, "available": True},
    "Dal Tadka":      {"category": "Mains",     "price": 180.0, "available": True},
    "Veg Biryani":    {"category": "Mains",     "price": 250.0, "available": True},
    "Garlic Naan":    {"category": "Mains",     "price":  40.0, "available": True},
    "Gulab Jamun":    {"category": "Desserts",  "price":  90.0, "available": True},
    "Rasgulla":       {"category": "Desserts",  "price":  80.0, "available": True},
    "Ice Cream":      {"category": "Desserts",  "price": 110.0, "available": False},
}

inventory = {
    "Paneer Tikka":   {"stock": 10, "reorder_level": 3},
    "Chicken Wings":  {"stock":  8, "reorder_level": 2},
    "Veg Soup":       {"stock": 15, "reorder_level": 5},
    "Butter Chicken": {"stock": 12, "reorder_level": 4},
    "Dal Tadka":      {"stock": 20, "reorder_level": 5},
    "Veg Biryani":    {"stock":  6, "reorder_level": 3},
    "Garlic Naan":    {"stock": 30, "reorder_level": 10},
    "Gulab Jamun":    {"stock":  5, "reorder_level": 2},
    "Rasgulla":       {"stock":  4, "reorder_level": 3},
    "Ice Cream":      {"stock":  7, "reorder_level": 4},
}

sales_log = {
    "2025-01-01": [
        {"order_id": 1,  "items": ["Paneer Tikka", "Garlic Naan"],          "total": 220.0},
        {"order_id": 2,  "items": ["Gulab Jamun", "Veg Soup"],              "total": 210.0},
        {"order_id": 3,  "items": ["Butter Chicken", "Garlic Naan"],        "total": 360.0},
    ],
    "2025-01-02": [
        {"order_id": 4,  "items": ["Dal Tadka", "Garlic Naan"],             "total": 220.0},
        {"order_id": 5,  "items": ["Veg Biryani", "Gulab Jamun"],           "total": 340.0},
    ],
    "2025-01-03": [
        {"order_id": 6,  "items": ["Paneer Tikka", "Rasgulla"],             "total": 260.0},
        {"order_id": 7,  "items": ["Butter Chicken", "Veg Biryani"],        "total": 570.0},
        {"order_id": 8,  "items": ["Garlic Naan", "Gulab Jamun"],           "total": 130.0},
    ],
    "2025-01-04": [
        {"order_id": 9,  "items": ["Dal Tadka", "Garlic Naan", "Rasgulla"], "total": 300.0},
        {"order_id": 10, "items": ["Paneer Tikka", "Gulab Jamun"],          "total": 270.0},
    ],
}


import copy

# ---------- Step 1: Deep copy ----------
inventory_backup = copy.deepcopy(inventory)

# ---------- Step 2: Prove deep copy ----------
inventory["Paneer Tikka"]["stock"] = 5

print("After temporary modification:")
print("Inventory  Paneer Tikka:", inventory["Paneer Tikka"])
print("Backup     Paneer Tikka:", inventory_backup["Paneer Tikka"])

# ---------- Step 3: Restore from backup ----------
inventory = copy.deepcopy(inventory_backup)

# ---------- Step 4: Final cart from Task 2 ----------
# Example: Paneer Tikka x3 in final cart
cart = [
    {"item": "Paneer Tikka", "quantity": 3, "price": 180.0}
]

# ---------- Step 5: Deduct stock based on cart ----------
for entry in cart:
    name = entry["item"]
    qty_needed = entry["quantity"]

    if name not in inventory:
        print(f"Item {name} not found in inventory.")
        continue

    current_stock = inventory[name]["stock"]

    if current_stock >= qty_needed:
        inventory[name]["stock"] = current_stock - qty_needed
    else:
        print(f"Warning: Only {current_stock} unit(s) available for {name}.")
        inventory[name]["stock"] = 0

# ---------- Step 6: Reorder alerts ----------
for item_name, details in inventory.items():
    if details["stock"] <= details["reorder_level"]:
        print(
            f"⚠ Reorder Alert: {item_name} — "
            f"Only {details['stock']} unit(s) left (reorder level: {details['reorder_level']})"
        )

# ---------- Step 7: Final prints ----------
print("\nFinal Inventory:")
print(inventory)

print("\nInventory Backup (unchanged):")
print(inventory_backup)
```

```python
# ----- Step 1: Revenue per day -----
print("Revenue per day:")

daily_revenue = {}  # store for reuse

for date, orders in sales_log.items():
    total_for_day = 0
    for order in orders:
        total_for_day += order["total"]
    daily_revenue[date] = total_for_day
    print(f"{date} → ₹{total_for_day:.2f}")

# ----- Step 2: Best‑selling day -----
best_day = None
best_revenue = 0.0

for date, amount in daily_revenue.items():
    if amount > best_revenue:
        best_revenue = amount
        best_day = date

print(f"\nBest-selling day: {best_day} ₹{best_revenue:.2f}")

# ----- Step 3: Most ordered item -----
item_count = {}

for orders in sales_log.values():
    for order in orders:
        for item in order["items"]:
            item_count[item] = item_count.get(item, 0) + 1

most_ordered_item = max(item_count, key=item_count.get)
print("Most ordered item:", most_ordered_item)

# ----- Step 4: Add new day -----
sales_log["2025-01-05"] = [
    {"order_id": 11, "items": ["Butter Chicken", "Gulab Jamun", "Garlic Naan"], "total": 490.0},
    {"order_id": 12, "items": ["Paneer Tikka", "Rasgulla"],                     "total": 260.0},
]

# ----- Step 5: Reprint revenue + new best‑selling day -----
print("\nUpdated Revenue per day:")

updated_daily_revenue = {}
for date, orders in sales_log.items():
    total_for_day = sum(order["total"] for order in orders)
    updated_daily_revenue[date] = total_for_day
    print(f"{date} → ₹{total_for_day:.2f}")

updated_best_day = None
updated_best_amount = 0.0

for date, amount in updated_daily_revenue.items():
    if amount > updated_best_amount:
        updated_best_amount = amount
        updated_best_day = date

print(f"Updated Best-selling day: {updated_best_day} ₹{updated_best_amount:.2f}")

# ----- Step 6: Numbered list of all orders (using enumerate) -----
print("\nAll Orders:")
all_orders = []

for date, orders in sales_log.items():
    for order in orders:
        all_orders.append((date, order))

for idx, (date, order) in enumerate(all_orders, start=1):
    items_str = ", ".join(order["items"])
    print(f"{idx}. [{date}] Order #{order['order_id']} — ₹{order['total']:.2f} — Items: {items_str}")
```
